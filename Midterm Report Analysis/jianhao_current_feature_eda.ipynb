## Analysis on Current Features - User Related
- Analyse `age`, `gender`, `is_subscriber`, `postcode`
- These describe the user profile

```
# Create working copy
beh = df_behaviors_train.copy()

# Number of candidate articles shown in each impression
beh["n_exposures"] = (
    beh["article_ids_inview"]
    .apply(len)
)

# Number of candidate articles clicked in each impression
beh["n_clicks"] = (
    beh["article_ids_clicked"]
    .apply(len)
)
```

```
user_fields = [
    "gender",
    "age",
    "is_subscriber",
    "postcode"
]

# One row per unique user
user_df = (
    df_behaviors_train[
        ["user_id"] + user_fields
    ]
    .drop_duplicates("user_id")
)

for col in user_fields:
    print(f"\n===== {col} =====")

    # Missingness
    missing_pct = (
        user_df[col]
        .isna()
        .mean()
        * 100
    )

    print(
        f"Missing: {missing_pct:.2f}%"
    )

    # Number of unique non-null values
    print(
        "Unique values:",
        user_df[col].nunique(
            dropna=True
        )
    )

    # Distribution
    print(
        user_df[col]
        .value_counts(
            dropna=False,
            normalize=True
        )
        .mul(100)
        .head(20)
    )
```

```
binary_cols = [
    "is_subscriber",
]

# One row per user
user_binary = (
    df_behaviors_train[
        ["user_id"] + binary_cols
    ]
    .drop_duplicates("user_id")
)

for col in binary_cols:
    print(f"\n===== {col} =====")

    dist = (
        user_binary[col]
        .value_counts(
            normalize=True,
            dropna=False
        )
        .mul(100)
    )

    print(dist)
```

```
for col in binary_cols:
    dist = (
        user_binary[col]
        .value_counts(normalize=True)
        .mul(100)
        .reset_index()
    )

    dist.columns = [
        col,
        "user_share_pct"
    ]

    plt.figure(figsize=(6, 4))

    plt.bar(
        dist[col].astype(str),
        dist["user_share_pct"]
    )

    plt.xlabel(col)
    plt.ylabel("Share of Users (%)")
    plt.title(
        f"Distribution of {col}"
    )

    plt.tight_layout()
    plt.show()
```

```
for col in binary_cols:
    ctr_table = (
        beh.groupby(col)
        .agg(
            exposures=(
                "n_exposures",
                "sum"
            ),
            clicks=(
                "n_clicks",
                "sum"
            )
        )
        .reset_index()
    )

    ctr_table["candidate_ctr"] = (
        ctr_table["clicks"]
        / ctr_table["exposures"]
    )

    print(f"\n===== {col} CTR =====")
    print(ctr_table)
```

```
overall_ctr = (
    beh["n_clicks"].sum()
    / beh["n_exposures"].sum()
)

# CTR by subscriber status
subscriber_ctr = (
    beh.groupby("is_subscriber")
    .agg(
        exposures=(
            "n_exposures",
            "sum"
        ),
        clicks=(
            "n_clicks",
            "sum"
        )
    )
    .reset_index()
)

subscriber_ctr["candidate_ctr"] = (
    subscriber_ctr["clicks"]
    / subscriber_ctr["exposures"]
)

# Unique user count by subscriber status
subscriber_users = (
    beh.groupby("is_subscriber")["user_id"]
    .nunique()
    .reset_index(name="user_count")
)

# Merge CTR and user counts
plot_df = subscriber_ctr.merge(
    subscriber_users,
    on="is_subscriber"
)

plot_df["status"] = (
    plot_df["is_subscriber"]
    .map({
        False: "Non-subscriber",
        True: "Subscriber"
    })
)

plot_df["ctr_pct"] = (
    plot_df["candidate_ctr"]
    * 100
)

# Plot
plt.figure(figsize=(7, 5))

bars = plt.bar(
    plot_df["status"],
    plot_df["ctr_pct"]
)

# Overall CTR reference line
plt.axhline(
    overall_ctr * 100,
    linestyle="--",
    label=f"Overall CTR ({overall_ctr * 100:.2f}%)"
)

# Add CTR + user count labels
for bar, (_, row) in zip(
    bars,
    plot_df.iterrows()
):
    height = bar.get_height()

    plt.text(
        bar.get_x()
        + bar.get_width() / 2,
        height + 0.03,
        (
            f"{row['ctr_pct']:.2f}%\n"
            f"n = {row['user_count']:,} users"
        ),
        ha="center",
        va="bottom"
    )

plt.xlabel("Subscriber Status")
plt.ylabel("Candidate CTR (%)")
plt.title(
    "Candidate CTR by Subscriber Status"
)

plt.legend()

# Give space for labels above bars
plt.ylim(
    0,
    plot_df["ctr_pct"].max() + 1
)

plt.tight_layout()
plt.show()
```

Findings
1. Age, Gender, Postcode have very limited coverage, with 98.12%, 95.73% and 98.61% of observations missing respectively. As a result, age-based, gender-based, postcode-based descriptive statistics and CTR comparisons would only reflect a small subset of users and may not be representative of the overall population. We therefore do not rely on age, gender, postcode as a primary model feature.
2. Subscriber status has complete coverage and is therefore retained for further analysis. Non-subscribers account for 96.74% of users, while subscribers account for only 3.26%, indicating a substantial class imbalance in this feature. Candidate CTR is slightly higher among non-subscribers at approximately 9.09%, compared with 8.79% among subscribers. The difference is relatively small, suggesting that subscriber status may have limited standalone predictive power. However, it is retained as a binary feature because it is fully observed and may provide additional value when combined with article-level features such as premium status.

```
# Create working copy
beh = df_behaviors_train.copy()

# Helper columns for candidate-level CTR
beh["n_exposures"] = (
    beh["article_ids_inview"]
    .apply(len)
)

beh["n_clicks"] = (
    beh["article_ids_clicked"]
    .apply(len)
)

# Derived context features
beh["is_front_page"] = (
    beh["article_id"]
    .isna()
)

beh["hour"] = (
    beh["impression_time"]
    .dt.hour
)

beh["day_of_week"] = (
    beh["impression_time"]
    .dt.dayofweek
)

weekday_map = {
    0: "Mon",
    1: "Tue",
    2: "Wed",
    3: "Thu",
    4: "Fri",
    5: "Sat",
    6: "Sun"
}

beh["weekday"] = (
    beh["day_of_week"]
    .map(weekday_map)
)


# =========================
# Distribution checks
# =========================

print("===== device_type =====")
print(
    beh["device_type"]
    .value_counts(
        dropna=False,
        normalize=True
    )
    .mul(100)
)

print(
    f"Missing: "
    f"{beh['device_type'].isna().mean() * 100:.2f}%"
)


print("\n===== is_sso_user =====")
print(
    beh["is_sso_user"]
    .value_counts(
        dropna=False,
        normalize=True
    )
    .mul(100)
)

print(
    f"Missing: "
    f"{beh['is_sso_user'].isna().mean() * 100:.2f}%"
)


print("\n===== page context =====")

page_dist = (
    beh["is_front_page"]
    .map({
        True: "Front page",
        False: "Article page"
    })
    .value_counts(
        normalize=True
    )
    .mul(100)
)

print(page_dist)


print("\n===== hour of day =====")
print(
    beh["hour"]
    .value_counts(
        normalize=True
    )
    .sort_index()
    .mul(100)
)


print("\n===== day of week =====")
print(
    beh["weekday"]
    .value_counts(
        normalize=True
    )
    .reindex([
        "Mon",
        "Tue",
        "Wed",
        "Thu",
        "Fri",
        "Sat",
        "Sun"
    ])
    .mul(100)
)
```

```
# Overall candidate CTR
overall_ctr = (
    beh["n_clicks"].sum()
    / beh["n_exposures"].sum()
)

# Build CTR tables
features = [
    "device_type",
    "is_sso_user",
    "is_front_page"
]

ctr_results = {}

for col in features:
    table = (
        beh.groupby(col)
        .agg(
            impressions=(
                "impression_id",
                "size"
            ),
            exposures=(
                "n_exposures",
                "sum"
            ),
            clicks=(
                "n_clicks",
                "sum"
            )
        )
        .reset_index()
    )

    table["candidate_ctr"] = (
        table["clicks"]
        / table["exposures"]
    )

    ctr_results[col] = table


# -------------------------
# Plot all 3 together
# -------------------------

fig, axes = plt.subplots(
    1,
    3,
    figsize=(16, 5)
)

plot_info = [
    (
        "device_type",
        "Device Type",
        None
    ),
    (
        "is_sso_user",
        "SSO Status",
        {
            False: "Non-SSO",
            True: "SSO"
        }
    ),
    (
        "is_front_page",
        "Page Context",
        {
            False: "Article page",
            True: "Front page"
        }
    )
]

for ax, (col, title, label_map) in zip(
    axes,
    plot_info
):
    df = ctr_results[col].copy()

    # Create readable x labels
    if label_map is not None:
        df["label"] = (
            df[col]
            .map(label_map)
        )
    else:
        df["label"] = (
            df[col]
            .astype(str)
        )

    # CTR in percentage points
    df["ctr_pct"] = (
        df["candidate_ctr"]
        * 100
    )

    bars = ax.bar(
        df["label"],
        df["ctr_pct"]
    )

    # Overall CTR reference
    ax.axhline(
        overall_ctr * 100,
        linestyle="--",
        label=(
            f"Overall CTR "
            f"({overall_ctr * 100:.2f}%)"
        )
    )

    # Annotate each bar
    for bar, (_, row) in zip(
        bars,
        df.iterrows()
    ):
        height = bar.get_height()

        ax.text(
            bar.get_x()
            + bar.get_width() / 2,
            height + 0.05,
            (
                f"{row['ctr_pct']:.2f}%\n"
                f"n={row['impressions']:,}"
            ),
            ha="center",
            va="bottom",
            fontsize=9
        )

    ax.set_title(title)
    ax.set_xlabel("")
    ax.set_ylabel(
        "Candidate CTR (%)"
    )

    ax.legend()

    # Give enough room for labels
    ax.set_ylim(
        0,
        df["ctr_pct"].max() + 1.3
    )

plt.suptitle(
    "Candidate CTR Across Impression Context Features",
    fontsize=14
)

plt.tight_layout()
plt.show()
```

## Analysis on Current Features - Historical Context
- Analyse History length from `article_id_fixed`, `read_time_fixed`, `scroll_percentage_fixed`
- These describe a user history

```
history = df_history_train.copy()


# -------------------------
# Helper function
# -------------------------

def safe_numeric_array(values):
    arr = np.asarray(
        values,
        dtype=float
    )

    return arr[
        np.isfinite(arr)
    ]


# -------------------------
# 1. History length
# -------------------------

history["history_length"] = (
    history["article_id_fixed"]
    .apply(len)
)


# -------------------------
# 2. Typical historical read time
# Use median because read time can be skewed
# -------------------------

history["median_read_time"] = (
    history["read_time_fixed"]
    .apply(
        lambda x:
        np.median(safe_numeric_array(x))
        if len(safe_numeric_array(x)) > 0
        else np.nan
    )
)


# -------------------------
# 3. Typical historical scroll percentage
# -------------------------

history["mean_scroll_percentage"] = (
    history["scroll_percentage_fixed"]
    .apply(
        lambda x:
        np.mean(safe_numeric_array(x))
        if len(safe_numeric_array(x)) > 0
        else np.nan
    )
)


# -------------------------
# Descriptive statistics
# -------------------------

history_features = [
    "history_length",
    "median_read_time",
    "mean_scroll_percentage"
]

for col in history_features:
    print(f"\n===== {col} =====")

    print(
        history[col]
        .describe(
            percentiles=[
                0.25,
                0.50,
                0.75,
                0.90,
                0.95,
                0.99
            ]
        )
    )

    print(
        f"Missing: "
        f"{history[col].isna().mean() * 100:.2f}%"
    )
```

```
import pandas as pd
import numpy as np

# =========================================
# 1. Aggregate behaviour to user level FIRST
# =========================================

user_beh = (
    beh.groupby(
        "user_id",
        as_index=False,
        sort=False
    )
    .agg(
        impressions=(
            "impression_id",
            "size"
        ),
        exposures=(
            "n_exposures",
            "sum"
        ),
        clicks=(
            "n_clicks",
            "sum"
        )
    )
)

print(user_beh.shape)
# Should be around 788k rows instead of 12m


# =========================================
# 2. Create groups on history table
# =========================================

history["history_length_group"] = pd.cut(
    history["history_length"],
    bins=[
        0,
        10,
        50,
        100,
        300,
        np.inf
    ],
    labels=[
        "0-9",
        "10-49",
        "50-99",
        "100-299",
        "300+"
    ],
    right=False
)

history["read_time_group"] = pd.qcut(
    history["median_read_time"],
    q=4,
    labels=[
        "Q1",
        "Q2",
        "Q3",
        "Q4"
    ],
    duplicates="drop"
)

history["scroll_group"] = pd.qcut(
    history["mean_scroll_percentage"],
    q=4,
    labels=[
        "Q1",
        "Q2",
        "Q3",
        "Q4"
    ],
    duplicates="drop"
)


# =========================================
# 3. Merge at USER level
# =========================================

user_analysis = user_beh.merge(
    history[
        [
            "user_id",
            "history_length_group",
            "read_time_group",
            "scroll_group"
        ]
    ],
    on="user_id",
    how="left"
)


# =========================================
# 4. Calculate CTR
# =========================================

def ctr_by_group(df, column):
    result = (
        df.groupby(
            column,
            observed=True
        )
        .agg(
            users=(
                "user_id",
                "size"
            ),
            impressions=(
                "impressions",
                "sum"
            ),
            exposures=(
                "exposures",
                "sum"
            ),
            clicks=(
                "clicks",
                "sum"
            )
        )
        .reset_index()
    )

    result["candidate_ctr"] = (
        result["clicks"]
        / result["exposures"]
    )

    return result


history_ctr = ctr_by_group(
    user_analysis,
    "history_length_group"
)

read_ctr = ctr_by_group(
    user_analysis,
    "read_time_group"
)

scroll_ctr = ctr_by_group(
    user_analysis,
    "scroll_group"
)


print("===== History Length CTR =====")
print(history_ctr)

print("\n===== Read Time CTR =====")
print(read_ctr)

print("\n===== Scroll Percentage CTR =====")
print(scroll_ctr)
```

```
import matplotlib.pyplot as plt

overall_ctr = (
    user_analysis["clicks"].sum()
    / user_analysis["exposures"].sum()
)

fig, axes = plt.subplots(
    1,
    3,
    figsize=(18, 5)
)

plot_info = [
    (
        history_ctr,
        "history_length_group",
        "History Length",
        "Prior Click History Length"
    ),
    (
        read_ctr,
        "read_time_group",
        "Historical Read Time",
        "Read Time Quartile"
    ),
    (
        scroll_ctr,
        "scroll_group",
        "Historical Scroll Percentage",
        "Scroll Depth Quartile"
    )
]

for ax, (
    df,
    group_col,
    title,
    xlabel
) in zip(
    axes,
    plot_info
):
    bars = ax.bar(
        df[group_col].astype(str),
        df["candidate_ctr"] * 100
    )

    # Overall CTR reference line
    ax.axhline(
        overall_ctr * 100,
        linestyle="--",
        label=(
            f"Overall CTR "
            f"({overall_ctr * 100:.2f}%)"
        )
    )

    # Add CTR + number of users
    for bar, (_, row) in zip(
        bars,
        df.iterrows()
    ):
        height = bar.get_height()

        ax.text(
            bar.get_x()
            + bar.get_width() / 2,
            height + 0.05,
            (
                f"{row['candidate_ctr'] * 100:.2f}%\n"
                f"n={row['users']:,}"
            ),
            ha="center",
            va="bottom",
            fontsize=8
        )

    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Candidate CTR (%)")
    ax.legend()

    ax.set_ylim(
        0,
        df["candidate_ctr"].max() * 100 + 1.5
    )

plt.suptitle(
    "Candidate CTR by Historical Behaviour Features",
    fontsize=14
)

plt.tight_layout()
plt.show()
```

Findings
1. Greater history length have higher CTR
2. Higher CTR for those with lesser read time and scroll percentage.

```
import pandas as pd
import gc

# Smaller = safer for RAM
chunk_size = 200_000

category_parts = []
article_type_parts = []
premium_parts = []


for start in range(
    0,
    len(df_behaviors_train),
    chunk_size
):
    end = min(
        start + chunk_size,
        len(df_behaviors_train)
    )

    # Only keep columns we actually need
    chunk = (
        df_behaviors_train
        .iloc[start:end][
            [
                "article_ids_inview",
                "article_ids_clicked"
            ]
        ]
        .copy()
    )

    # Convert clicked list to set
    # for faster membership checking
    chunk["clicked_set"] = (
        chunk["article_ids_clicked"]
        .apply(set)
    )

    # One row per candidate article
    chunk = (
        chunk
        .explode("article_ids_inview")
        .rename(
            columns={
                "article_ids_inview":
                "article_id"
            }
        )
    )

    # Candidate click label
    chunk["clicked"] = [
        int(article_id in clicked_set)
        for article_id, clicked_set
        in zip(
            chunk["article_id"],
            chunk["clicked_set"]
        )
    ]

    # Remove large columns ASAP
    chunk = chunk.drop(
        columns=[
            "article_ids_clicked",
            "clicked_set"
        ]
    )

    # Add article features
    chunk = chunk.merge(
        df_articles[
            [
                "article_id",
                "category",
                "article_type",
                "premium"
            ]
        ],
        on="article_id",
        how="left"
    )

    # -------------------------
    # Category
    # -------------------------
    category_chunk = (
        chunk.groupby(
            "category",
            dropna=False
        )
        .agg(
            exposures=(
                "clicked",
                "size"
            ),
            clicks=(
                "clicked",
                "sum"
            )
        )
        .reset_index()
    )

    category_parts.append(
        category_chunk
    )

    # -------------------------
    # Article type
    # -------------------------
    article_type_chunk = (
        chunk.groupby(
            "article_type",
            dropna=False
        )
        .agg(
            exposures=(
                "clicked",
                "size"
            ),
            clicks=(
                "clicked",
                "sum"
            )
        )
        .reset_index()
    )

    article_type_parts.append(
        article_type_chunk
    )

    # -------------------------
    # Premium
    # -------------------------
    premium_chunk = (
        chunk.groupby(
            "premium",
            dropna=False
        )
        .agg(
            exposures=(
                "clicked",
                "size"
            ),
            clicks=(
                "clicked",
                "sum"
            )
        )
        .reset_index()
    )

    premium_parts.append(
        premium_chunk
    )

    # Clear current chunk from memory
    del chunk
    gc.collect()

    print(
        f"Processed "
        f"{end:,} / "
        f"{len(df_behaviors_train):,}"
    )


# =========================================
# Combine chunk results
# =========================================

def combine_results(
    parts,
    feature
):
    result = (
        pd.concat(
            parts,
            ignore_index=True
        )
        .groupby(
            feature,
            dropna=False,
            as_index=False
        )[
            [
                "exposures",
                "clicks"
            ]
        ]
        .sum()
    )

    result["candidate_ctr"] = (
        result["clicks"]
        / result["exposures"]
    )

    result["exposure_share_pct"] = (
        result["exposures"]
        / result["exposures"].sum()
        * 100
    )

    return result.sort_values(
        "exposures",
        ascending=False
    )


category_ctr = combine_results(
    category_parts,
    "category"
)

article_type_ctr = combine_results(
    article_type_parts,
    "article_type"
)

premium_ctr = combine_results(
    premium_parts,
    "premium"
)


# =========================================
# Results
# =========================================

print("\n===== CATEGORY CTR =====")
print(category_ctr)

print("\n===== ARTICLE TYPE CTR =====")
print(article_type_ctr)

print("\n===== PREMIUM CTR =====")
print(premium_ctr)
```
